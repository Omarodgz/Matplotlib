## Ejercicio 1:  Datos de la estación meteorologica de Progreso, Yucatán

Fuente:  https://smn.conagua.gob.mx/es/observando-el-tiempo/estaciones-meteorologicas-automaticas-ema-s

```python
df_clima = pd.read_csv("datos/Estacion_PROGRESO_90_dias_19Nov.csv")
df_clima.head()
```

<div class="alert alert-success">

1. Considerando todos los datos hacer una grafica de dispersión para la Temperatura de Aire (en el eje Y) y la Humadad Relativa (en el eje X) 
</ul>
</div>

```python
# Estilo Matlab
data_TA = df_clima['TemperaturaAire(C)']
data_HR = df_clima['HumedadRelativa(%)']

X = data_HR
Y = data_TA

plt.figure(figsize=(6,3))  
plt.scatter(X, Y, c= 'red' , s = 40 , marker='X')

plt.title('Temperatura de Aire (C) vs. Humedad Relativa (%)',fontsize=10)
plt.xlabel('Humedad Relativa (%)',fontsize=10)
plt.ylabel('Temperatura de Aire (C)',fontsize=10);
plt.show()
```

# Ejercicio 2: Incendios en México

```python
df_incendios = pd.read_csv("datos/Serie_historica_anual_incendios_2017.csv")
df_incendios.head()
```

<div class="alert alert-success">
    
1. Utilizar una grafica de barras para comparar el total de hectáreas afectadas para los 5 estados más afectados 
    
</div> 

```python
# Estilo Matlab
Temporal = df_incendios.groupby('Estado')['Total Hectareas'].sum().reset_index().sort_values('Total Hectareas',ascending=False).head(5)

x_valores = Temporal['Estado']
y_valores = Temporal['Total Hectareas']

plt.figure(figsize=(8,4))  
plt.bar(x_valores, y_valores)       

plt.title('Total de hectareas afectadas para los 5 estados más dañados',fontsize=12)
plt.xlabel('Estado',fontsize=10)
plt.ylabel('Cantidad de hectareas afectadas',fontsize=10);
plt.show()
```

```python
# Estilo Orientado a Objetos
fig, ax = plt.subplots(figsize=(8,4), dpi=100)

ax.bar(x_valores, y_valores)          
ax.set(title='Total de hectareas afectadas para los 5 estados más dañados', xlabel= 'Estado' , ylabel='Cantidad de hectareas afectadas')
plt.show()
```

<div class="alert alert-success">
    
2. Utilizar una grafica de barras para comparar el total de hectáreas afectadas para Yucatán para los meses: abril, mayo, junio, julio y agosto 
    
</div> 

```python
# Estilo Matlab
str_estado = 'Yucatán'
str_abril = 'abril'
str_mayo = 'mayo'
str_junio = 'junio'
str_julio = 'julio'
str_agosto = 'agosto'

mascara_estado = df_incendios['Estado'].str.contains(str_estado) 
mascara_abril = df_incendios['Mes Acontecimiento'].str.contains(str_abril)
mascara_mayo = df_incendios['Mes Acontecimiento'].str.contains(str_mayo)
mascara_junio = df_incendios['Mes Acontecimiento'].str.contains(str_junio)
mascara_julio = df_incendios['Mes Acontecimiento'].str.contains(str_julio)
mascara_agosto = df_incendios['Mes Acontecimiento'].str.contains(str_agosto)

total_abril = df_incendios[(mascara_estado) & (mascara_abril) ]['Total Hectareas'].sum() 
total_mayo = df_incendios[(mascara_estado) & (mascara_mayo) ]['Total Hectareas'].sum()
total_junio = df_incendios[(mascara_estado) & (mascara_junio) ]['Total Hectareas'].sum()
total_julio = df_incendios[(mascara_estado) & (mascara_julio) ]['Total Hectareas'].sum()
total_agosto = df_incendios[(mascara_estado) & (mascara_agosto) ]['Total Hectareas'].sum() 

x_valores2 = [ str_abril, str_mayo, str_junio, str_julio, str_agosto  ]
y_valores2 = [  total_abril, total_mayo, total_junio , total_julio , total_agosto  ]

plt.figure(figsize=(6,4))  
plt.bar(x_valores2, y_valores2)          
plt.title('Total de hectareas afectadas para Yucatán por mes',fontsize=12)
plt.xlabel('Mes',fontsize=10)
plt.ylabel('Cantidad de hectareas afectadas',fontsize=10);
plt.show()
```

```python
# Estilo Orientado a Objetos
fig, ax = plt.subplots(figsize=(6,4), dpi=100)

ax.bar(x_valores2, y_valores2)          
ax.set(title='Total de hectareas afectadas para Yucatán por mes', xlabel= 'Mes' , ylabel='Cantidad de hectareas afectadas')
plt.show()
```

<div class="alert alert-success">

```python
# Estilo Orientado a Objetos
fig, ax = plt.subplots(figsize=(6,5), dpi=100)

ax.hist(datos_D,10, density=False, facecolor='blue')
ax.set(title='Histograma de valores de duracion de los incendios', xlabel= 'Valores de duracion de los incendios (horas decimales)', ylabel='Frecuencia')
ax.grid(False)

plt.show()
```

<div class="alert alert-success">
    
4. Considerando todos los datos, hacer una grafica de dispersión entre la duración (eje X) y el total de hectáreas (eje Y)
    
</div> 

```python
# Estilo Matlab
aux = df_incendios[df_incendios['Duracion H'] < 24]

data_D = aux['Duracion H']
data_TH = aux['Total Hectareas']

X = data_D
Y = data_TH

plt.figure(figsize=(6,4))  
plt.scatter(X, Y, c= 'blue' , s = 40 , marker='X')
plt.title('Duracion vs. Total de hectareas',fontsize=12)
plt.xlabel('Duracion en horas decimales',fontsize=10)
plt.ylabel('Cantidad total de hectareas afectadas',fontsize=10);
plt.show()
```

# Ejercicio 3: Pacientes confirmados con Covid en México


Fuente: https://datos.covid-19.conacyt.mx/#DownZCSV

```python
df_covid = pd.read_csv("datos/Casos_Diarios_Estado_Nacional_Confirmados_20201129.csv")
```

```python
df_covid = df_covid.loc[:,'nombre':]
df_covid.head()
```

```python
df_covid = df_covid.set_index('nombre').T
df_covid.head()
```

```python
df_covid = df_covid.reset_index()
df_covid.head()
```

```python
df_covid = df_covid.rename(columns = {'index': 'Fecha'})
df_covid.head()
```

```python
df_covid['Fecha'] = pd.to_datetime(df_covid['Fecha'], format= '%d-%m-%Y')
df_covid = df_covid.set_index('Fecha')
df_covid
```

<div class="alert alert-success">
    
1. Hacer una grafica de barras comparando todos los meses, considerando el total nacional (por mes) 
    
</div> 

```python
# Estilo Matlab
import calendar

temporal = df_covid['Nacional'].resample('M').sum()
temporal2 = pd.DataFrame(temporal)
temporal3 = temporal2.reset_index()
temporal4 = temporal3.rename(columns = {'Nacional':'Total nacional'})
temporal4['Ext'] = pd.DatetimeIndex(temporal4['Fecha']).month

def month(x):
    return calendar.month_name[int(x)] if not np.isnan(x) else np.nan

temporal4['Mes'] = temporal4['Ext'].apply(month)

x_val = temporal4['Mes']
y_val = temporal4['Total nacional']

plt.figure(figsize=(12,4))  
plt.bar(x_val, y_val,color='green')          
plt.title('Total de casos confirmados a nivel nacional por mes',fontsize=12)
plt.xlabel('Mes',fontsize=10)
plt.ylabel('Cantidad de casos confirmados',fontsize=10);

plt.show()
```

```python
# Estilo Orientado a Objetos
fig, ax = plt.subplots(figsize=(12,4), dpi=100)

ax.bar(x_val, y_val,color='green')          
ax.set(title='Total de casos confirmados a nivel nacional por mes', xlabel= 'Mes' , ylabel='Cantidad de casos confirmados')

plt.show()
```

<div class="alert alert-success">
    
2. Hacer una grafica de barras comparando  los meses de agosto, septiembre, octubre y noviembre, considerando solo las entidades de Yucatán, Distrito Federal  y Jalisco
    
</div> 

```python
# Estilo Matlab
import calendar

tempYUC = df_covid['YUCATAN'].resample('M').sum()
tempDF = df_covid['DISTRITO FEDERAL'].resample('M').sum()
tempJA = df_covid['JALISCO'].resample('M').sum()
temp2 = pd.DataFrame({'Yucatán':tempYUC,'Distrito Federal':tempDF,'Jalisco':tempJA})
temp3 = temp2.reset_index()
temp3['Ext'] = pd.DatetimeIndex(temp3['Fecha']).month

def month(x):
    return calendar.month_name[int(x)] if not np.isnan(x) else np.nan

temp3['Mes'] = temp3['Ext'].apply(month)

def suma(x,y,z):
    return x+y+z

temp3['Total tres estados'] = temp3.apply(lambda x: suma(x['Yucatán'], x['Distrito Federal'], x['Jalisco']),axis=1,result_type='expand')

temp4 = temp3.tail(4)

x_valor = temp4['Mes']
y_valor = temp4['Total tres estados']

plt.figure(figsize=(7,4))  
plt.bar(x_valor, y_valor,color='green')          
plt.title('Total de casos confirmados en las entidades de Yucatán, Distriro Federal y Jalisco por mes',fontsize=12)
plt.xlabel('Mes',fontsize=10)
plt.ylabel('Cantidad de casos confirmados',fontsize=10);

plt.show()
```

```python
# Estilo Orientado a Objetos
fig, ax = plt.subplots(figsize=(7,4), dpi=100)

ax.bar(x_valor, y_valor,color='green')          
ax.set(title='Total de casos confirmados en las entidades de Yucatán, Distriro Federal y Jalisco por mes', xlabel= 'Mes' , ylabel='Cantidad de casos confirmados')
plt.show()
```
